STRINGS CONCATINATION --> String concatenation means joining two or more strings together to form a single string with + sign.

```python
hey = "Hello " + "Yashoda"
print(hey)
```

STRINGS REPLICATION --> String replication means repeating a string multiple times with * sign.

```python
hey = "Hello " * 5
print(hey)
```

TYPE CASTING =Type casting means converting a value from one data type to another data type manually.

```python
x = "10"
print(type(x))
y = int(x)
print(type(y))
```

TYPE CONVERSION = Type conversion is when the programming language automatically converts one data type into another.

```python
a = 2
b = 5
print(type(a))
print(type(b))
c = a / b
print(c)
print(type(c))
```
