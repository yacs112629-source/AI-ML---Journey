#OPERATORS
✅1. Arithmetic operators --> Used to perform mathematical calculations.

| Operator | Meaning             | Example   | Output |
| -------- | ------------------- | --------- | -----: |
| `+`      | Addition            | `10 + 5`  |   `15` |
| `-`      | Subtraction         | `10 - 5`  |    `5` |
| `*`      | Multiplication      | `10 * 5`  |   `50` |
| `/`      | Division            | `10 / 5`  |  `2.0` |
| `%`      | Modulus (remainder) | `10 % 3`  |    `1` |
| `//`     | Floor division      | `10 // 3` |    `3` |
| `**`     | Exponent/Power      | `2 ** 3`  |    `8` |

✅2. Comparison operator --> Used to compare two values. The result is always True or False.

| Operator | Meaning                  | Example  | Output  |
| -------- | ------------------------ | -------- | ------- |
| `==`     | Equal to                 | `5 == 5` | `True`  |
| `!=`     | Not equal to             | `5 != 3` | `True`  |
| `>`      | Greater than             | `5 > 3`  | `True`  |
| `<`      | Less than                | `5 < 3`  | `False` |
| `>=`     | Greater than or equal to | `5 >= 5` | `True`  |
| `<=`     | Less than or equal to    | `3 <= 5` | `True`  |

✅3. Assignment operator --> Used to assign values to variables.

| Operator | Example  | Same as     |
| -------- | -------- | ----------- |
| `=`      | `x = 5`  | `x = 5`     |
| `+=`     | `x += 2` | `x = x + 2` |
| `-=`     | `x -= 2` | `x = x - 2` |
| `*=`     | `x *= 2` | `x = x * 2` |
| `/=`     | `x /= 2` | `x = x / 2` |
| `%=`     | `x %= 2` | `x = x % 2` |

✅4. Lgical operator --> Used to combine conditions.

4.1. and -->
Returns True if both conditions are true otherwise False.

4.2. or -->
Returns False if both conditions are false otherwise True.

4.3. not -->
Reverses the result.

✅5. Mebership operator -->
Used to check whether a value exists inside a sequence like a string, list, etc.

5.1. in -->
Checks whether a value is present.
example - fruits = ["apple", "banana", "mango"]

print("apple" in fruits)

5.2. not in -->
Checks whether a value is not present.
example - fruits = ["apple", "banana", "mango"]

print("orange" not in fruits)
