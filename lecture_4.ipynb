#IF QUESTIONS

Question: Given the age of the person chk if they are eligible to vote . if age is atleast 18, then they are eligible

```python
age = int(input("Enter Your Age ==> "))
if age >= 18:
  print("Yes you are eligible to vote")
```

Question: Take the password from the user. if it matches "hello@123" then print "Welcome"

```python
name = input("What is you name ? ")
password = input("Tell me your password ? ")
if password == "hello@123":
  print("Welcome",name)
```

#IF-ELSE QUESTIONS

```python
age = int(input("Enter Your Age ==> "))
if age >= 18:
  print("Yes you are eligable vote")
else:
  print("No you are not eligible to vote")
```

```python
name = input("What is you name ? ")
password = input("Tell me your password ? ")
if password == "hello@123":
  print("Welcome",name)
else:
  print("Invalid Password")
```

#IF-ELIF-ELSE QUESTIONS

```python
marks = int(input("Tell me your marks = "))
if marks < 30:
  print("Low")
elif marks >= 30 and marks <= 70:
  print("Average")
else:
  print("High")
```
